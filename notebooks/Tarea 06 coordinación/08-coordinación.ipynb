## 2.2 Diagrama de Gantt — Timeline del proyecto

```python
# ============================================================
# GANTT CHART con px.timeline
# ============================================================

df_gantt = tareas.copy()
df_gantt['Inicio'] = pd.to_datetime(df_gantt['Inicio'])
df_gantt['Fin']    = pd.to_datetime(df_gantt['Fin'])

color_map = {
    '✅ Completado': '#66BB6A',
    '🔄 En curso':   '#FFA726',
}

fig_gantt = px.timeline(
    df_gantt,
    x_start='Inicio',
    x_end='Fin',
    y='Tarea',
    color='Estado',
    color_discrete_map=color_map,
    hover_data=['Solicitante', 'Prioridad', 'Descripción'],
    title='Plan de Proyecto — easyMoney Data Science (jun–jul 2019)'
)

fig_gantt.update_yaxes(autorange='reversed')
fig_gantt.update_layout(
    template='plotly_white',
    height=380,
    width=950,
    margin=dict(l=220),
    xaxis=dict(tickformat='%d %b', tickangle=-30, title='Fecha'),
    legend=dict(title='Estado', x=1.01, y=0.5)
)

fig_gantt.show()
```
