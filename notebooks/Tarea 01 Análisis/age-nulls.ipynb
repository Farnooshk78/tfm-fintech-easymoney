```python
null_salary = df_socio[df_socio['salary'].isna()].copy()
total_nulls = len(null_salary)

print(f"Total nulls en salary: {total_nulls:,}")
print(f"\nDistribución por edad (únicos por pk_cid):")

age_dist = null_salary.groupby('age')['pk_cid'].nunique().reset_index()
age_dist.columns = ['age', 'clientes_unicos']
age_dist['pct'] = (age_dist['clientes_unicos'] / age_dist['clientes_unicos'].sum() * 100).round(2)
age_dist = age_dist.sort_values('age')
print(age_dist.to_string())

print(f"\nResumen por grupos:")
print(f"  Edad < 20:    {null_salary[null_salary['age'] < 20]['pk_cid'].nunique():>8,} clientes únicos")
print(f"  Edad 20-24:   {null_salary[null_salary['age'].between(20,24)]['pk_cid'].nunique():>8,} clientes únicos")
print(f"  Edad 25-64:   {null_salary[null_salary['age'].between(25,64)]['pk_cid'].nunique():>8,} clientes únicos")
print(f"  Edad 65+:     {null_salary[null_salary['age'] >= 65]['pk_cid'].nunique():>8,} clientes únicos")
```
